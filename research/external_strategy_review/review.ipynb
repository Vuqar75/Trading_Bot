# External strategy receipt — offline companion
Reads public pinned repository snapshots as data; does not import or run their trading code.
Set the three checkout paths below to the exact commits listed in the report.
The complete assertions and arithmetic are in tools/review_external_strategy_ledgers.py.


```python
from pathlib import Path
from types import SimpleNamespace
import importlib.util
root = Path.cwd()
while not (root / "tools/review_external_strategy_ledgers.py").exists():
    if root == root.parent: raise FileNotFoundError("Open inside Trading_Bot")
    root = root.parent
spec = importlib.util.spec_from_file_location("ledger_review", root / "tools/review_external_strategy_ledgers.py")
review = importlib.util.module_from_spec(spec)
spec.loader.exec_module(review)
paths = SimpleNamespace(
    josiah=Path("/workspace/work/external-crypto-bot-review"),
    rl=Path("/workspace/work/external-rl-review"),
    sentinel=Path("/workspace/work/external-sentinel-review"),
)
receipt = review.run(paths)
```
